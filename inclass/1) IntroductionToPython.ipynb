#Function
**Step1: Create a function with def keyword**

def functionName(parameter1, parameter2, ....):


> Function Body


> return result

**Step2: Call the function to execute it**

functionName(argument1, argument2, ....)

**Points to remember:**
1. A function can have any number of parameters. It may not have any paramemeter also.
2. A python function can return any number of results at the same time. It may not return any result also. (This is different from C++/Java)
3. When a function is called, values of the arguments replace the parameters in the function in order (i.e. argument1 replaces parameter1, argument 2 replaces parameter2, and so on)
4. A function can be written anywhere in the code



```python
#Step1: Create a function
def addTwoNumbers(n1, n2): #n1 and n2 are the parameters.
  result = n1 + n2 #Function Body
  return result #Return the result to where the function is called

#Step2: Call the function to execute it
returnedResult = addTwoNumbers(5, 4.5) #Returned result is stored in a variable
print(returnedResult)
```

```python
#A function with no return

#Step1: Create a function
def printMsg(msgTxt):
  print(msgTxt)
  return #Optional.
         #However, it is a good practice to write return
         #even when nothing is returning actually

#Step2: Call the function
printMsg("Welcome to UAF") #Since the function is not returning anything
                           #you cannot store it in a variable
```

```python
#A Python function that returns multiple values
#Step1: Create a function
def doArithmeticOnTwoNumbers(n1, n2):
  resAdd = n1 + n2
  resSub = n1 - n2
  resMul = n1 * n2
  resDiv = n1 / n2
  return (resAdd, resSub, resMul, resDiv)

#Step2: Call the function
(retResAdd, retResSub, retResMul, retResDiv) = doArithmeticOnTwoNumbers(4, 2)
#All results are stored in order of their return.
print(retResAdd)
print(retResSub)
print(retResMul)
print(retResDiv)
```
