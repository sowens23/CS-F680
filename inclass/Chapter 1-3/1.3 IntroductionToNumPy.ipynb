#Why NumPy?

1. Lots of inbuilt operators/functions (Less lines of codes)  
2. Faster
3. Memory efficient



```python
#List Vs NumPy (Which one is faster?)
import time

size_of_vec = 1000000

t1 = time.time()
X = range(size_of_vec)
Y = range(size_of_vec)
for i in range(0, len(X), 1):
    Z = X[i] + Y[i] #This is another way of writing python loop 
time_taken_regularLoop = time.time() - t1
print("Time taken by regular loop = ", time_taken_regularLoop)

t1 = time.time()
X = np.arange(size_of_vec)
Y = np.arange(size_of_vec)
Z = X + Y
time_taken_numpy = time.time() - t1
print("Time taken by numpy = ", time_taken_numpy)

print("Numpy in this example " + str(time_taken_regularLoop/time_taken_numpy) + " faster!")
#More performance gain for bigger data
```

#Broadcasting concept
1. Fancy name for using Universal Function (Function that operates element wise on arrays)
2. A smaller dimentional array can be broadcasted to the larger dimentional arra
3. But there is a rule: The size of all the trailing dimension of both the arrays must be either EQUAL or ONE (ONE means you can just strech it to match the larger dimentional array).

Note: It becomes pretty complicated at higher dimension

```python
#What happen when you do an arithmetic operation 
#with an array (vector) and a regular variable (scaler)
arr_1d = np.array([1, 2, 3]) #1*3
var = 2 #1*1
result = arr_1d + var #var => var = np.array(var, var, var)
#result
```

```python
#Extend that concept to multiple dimention
#A one dimensional array added to a two dimensional array 
#. results in broadcasting if number of 1-d array 
#. elements matches the number of 2-d array columns.
arr_2d = np.array([[1, 2, 3], [4, 5, 6]]) #shape: 2*3
arr_1d = np.array([1, 2, 3]) #shape: 1*3
result = arr_2d + arr_1d #arr_1d is broadcasted to arr_2d: 
                         #arr_1d added one extra row (copy of the first row) 
result
```

#Querying an array

```python
#Load Data from File
#filedata = np.genfromtxt('data.txt', delimiter=',') #From local machine

filedata=np.array([[1,13,21,11,196,75,4,3,34,6,7,8,0,1,2,3,4,5],
                  [3,42,12,33,766,75,4,55,6,4,3,4,5,6,7,0,11,12],
                  [1,22,33,11,999,11,2,1,78,0,1,2,9,8,7,1,76,88]])

filedata = filedata.astype('int32')
print(filedata)
```

```python
#Boolean condition checking on each cell of the entire array
result = filedata[(filedata > 50)]
print(result)

result = filedata[(filedata > 50) & (filedata<90)]
print(result)
```

```python
#Boolean condition checking on each cell of a specific row
row = 0
result = filedata[row, (filedata[row] > 50)]
print(result)

result = filedata[row, (filedata[row] > 50) & (filedata[row] < 90)]
print(result)
```
