#Dataframe and Series: Pandas fundamental datastructure

```python
#Create a dictionry
data = {
    'id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Cindy', 'David', 'Elizabeth']
}
type(data)
```

```python
#Read the dictionary into pandas
df = pd.DataFrame(data) #Read the entire dictionary with two columns (id and name) as a Dataframe
df
srs = pd.Series(data['id'])
print(type(srs))
srs
#print(type(df['id'])) #Each column of a dataframe is also a series
```

```python
#Convert a numpy array to Pandas and viceversa

import numpy as np
my_array = np.array([[11,22,33],
                     [44,55,66]])#Creating a numpy array

df = pd.DataFrame(my_array, columns = ['Column_A','Column_B','Column_C']) #convert from numpy to Pandas

print(df)
print(type(df))

narray = df.to_numpy() #convert from  Pandas to numpy
print(narray)
print(type(narray))
```

#Preprocess the data

```python
#Read a file available over internet (Copy-paste the previous block so that I don't need to scroll too much)
#In this section, I will make some changes in my file and you should be able to see those changes in your program)
file_url = 'https://docs.google.com/spreadsheets/d/e/2PACX-1vQL0mu4CnfA3o4cE_aXdkVIgb3x28aV1ayVohgpBw4EKYV0DmUFr5mcuXRyYvoUUjXXJnc6e9Mabp3f/pub?output=csv'
df_cancer = pd.read_csv(file_url)
#type(df_cancer)
df_cancer
```

```python
#Remove duplicates
df_cancer.drop_duplicates(inplace=True) #If you do not use inplace=True it will create another copy of the dataframe with no duplicate rows
print(sum(df_cancer.duplicated()))
```

```python
#Remove null values
df_cancer.dropna(inplace=True, axis=0) #If you do not use inplace=True it will create another copy of the dataframe with no null values
df_cancer
```

```python
#Renaming columns (This may impact following blocks)
print(df_cancer.columns)
df_cancer.rename(columns={
    '%id': 'id',
    'diagnosis (M=malignant; B=benign)': 'diagnosis'
}, inplace=True)
print(df_cancer.columns)
```

#Filtering based on some statistics
Let us mix both filter/serach/query and maths/stats

```python
#DIY: Get the '%id' of those recods where area is less than the mean(area) but greater than the min(area)

display_column_name1 = 'id' #Change it to id if you have changed it earlier
analytic_column_name2 = 'area (nucA)'
df_cancer[ (df_cancer[analytic_column_name2] > df_cancer[analytic_column_name2].min()) & (df_cancer[analytic_column_name2] < df_cancer[analytic_column_name2].mean()) ] [[display_column_name1, analytic_column_name2]]
```

#Group By

```python
#Read a file available over internet (Copy-paste the previous block so that I don't need to scroll too much)
#In this section, I will make some changes in my file and you should be able to see those changes in your program)
file_url = 'https://docs.google.com/spreadsheets/d/e/2PACX-1vSxuqWA__mPgb5p0abdX5MsV64mijIs_wblVo6j-H3vuSLgG8c_RirNMc1zHX_FCXdWgBgbv0VtCNgq/pub?output=csv'
df_cancer = pd.read_csv(file_url)
#type(df_cancer)
df_cancer
```

```python
#You can group by based on as many columns as you want in comma separated fashion
#For one column you may not need to use the [] but that will not work for mutiple columns 
grouped_df_cancer = df_cancer.groupby(['diagnosis (M=malignant; B=benign)']) 
#for key, item in grouped_df_cancer: 
#    print(grouped_df_cancer.get_group(key)['%id'], "\n\n")
grouped_df_cancer
```

```python
grouped_df_cancer = df_cancer.groupby(['diagnosis (M=malignant; B=benign)'])
grouped_df_cancer.head(4)
```

```python
#See the grouped statistics for each column
grouped_df_cancer = df_cancer.groupby(['diagnosis (M=malignant; B=benign)']).agg('sum')
grouped_df_cancer
```
