#List: [ ]
Use the same name followed by an index to access a data element. Similar to calling the nth person in a queue as person[n] instead of calling him/her by name say Alice.

```python
person = ["Alice", "Bob", "Cindy", "David"]
print(type(person))
print(person[2])
```

```python
#Python can store multiple different types of elements in a list
my_list = ["Alice", 22, 900.45, True, "Alice"]
print(my_list)
print(type(my_list))
print(type(my_list[0]))
print(type(my_list[1]))
print(type(my_list[2]))
print(type(my_list[3]))
```

```python
#Modify a list element
my_list = ["Alice", 22, 900.45, True]
print(my_list)
my_list[3] = "Bob"
print(my_list)
print(my_list[0])
```

#Set: { }
Unordered list with unique values

CanNOT be accessed using numerical index 

```python
my_set = {"Alice", 22, 900.45, True}
print(my_set)

#print(my_set[0]) #Error: set element cannot be accessed using index
```

```python
#Set automatically eleminates duplicates (each elements are unique)
my_list = ["Alice", 22, 900.45, True]
my_tuple = ("Alice", 22, 900.45, True)
my_set = {"Alice", 22, 900.45, True}

my_list.append("Alice") #add a duplicated element to my_list
print(my_list)

#my_tuple.append("Alice") #Error: Tuple is immutable

my_set.add("Alice") #add a duplicated element to my_set
print(my_set)
```

#Dictionary
1. List with user defined index (key)
2. Stores key: value pair
3. Keys are unique (Duplicates are eliminated automatically)
4. Another way of viewing a dictionary as an extended version of set 

```python
my_dict = {'Name': "Alice",
'Age': 22,
'Salary': 900.45,
'isStudent': True
}
print(my_dict['Name'])
print(my_dict['Age'])
print(my_dict['Salary'])
print(my_dict['isStudent'])
```
